```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon
import numpy as np


# Load trained model

model = YOLO("bee_mite_segmentation/yolov11_seg_bee_a100_fast/weights/best.pt")

# Predict on image

image_path = "/content/test_images/WhatsApp Image 2026-01-08 at 11.43.56 AM.jpeg"
results = model.predict(source=image_path, conf=0.25)


# Get image and masks

orig_img = results[0].orig_img  # original image as numpy array
masks = results[0].masks        # segmentation masks
names = results[0].names        # class names
classes_detected = [names[int(cls)] for cls in results[0].boxes.cls]

print("Number of masks detected:", len(masks.xy))
print("Detected classes:", classes_detected)


# Display only polygons (masks) on the image safely

plt.figure(figsize=(10, 10))
plt.imshow(orig_img)
plt.axis("off")

# Loop over each detected mask
for mask_xy in masks.xy:

    if isinstance(mask_xy[0][0], (list, tuple, np.ndarray)):  # multiple polygons
        polygons = mask_xy
    else:  # single polygon
        polygons = [mask_xy]

    for poly in polygons:
        poly = np.array(poly)
        if poly.shape[1] != 2:
            continue  # skip invalid polygons
        plt.gca().add_patch(
            Polygon(poly, fill=False, edgecolor='red', linewidth=2)
        )

plt.show()

```
